# 05 — XGBoost (Gradient Boosting)
Objectif : Utiliser l'état de l'art pour les données tabulaires.

```
import pandas as pd
import xgboost as xgb
from sklearn.metrics import accuracy_score
from sklearn.preprocessing import LabelEncoder

df = pd.read_csv('../features/ml_dataset.csv').dropna(subset=['result'])
features = ['home_form_5', 'away_form_5', 'home_avg_overall', 'away_avg_overall', 'home_squad_value', 'away_squad_value', 'odds_prob_home', 'odds_prob_draw', 'odds_prob_away']
X = df[features].fillna(df[features].median())
y = LabelEncoder().fit_transform(df['result'])

X_train, y_train = X[df['season'] != '2024/25'], y[df['season'] != '2024/25']
X_test, y_test = X[df['season'] == '2024/25'], y[df['season'] == '2024/25']

model = xgb.XGBClassifier(n_estimators=100, learning_rate=0.1, random_state=42)
model.fit(X_train, y_train)
print(f"Accuracy XGBoost Baseline: {accuracy_score(y_test, model.predict(X_test)):.2%}")
```
